```python
import sqlite3
import pandas as pd
from pathlib import Path
csv_path = Path("/Users/keerthanavenkatesan/Documents/Data Managment/CS210FinalProject/Spotify_Popularity_Prediction/data/raw/spotify_100k.csv")
df = pd.read_csv(csv_path)

print(df.shape)
df.head()
```

```python
db_path = Path("/Users/keerthanavenkatesan/Documents/Data Managment/CS210FinalProject/Spotify_Popularity_Prediction/data/spotify.db")

conn = sqlite3.connect(db_path)
df.to_sql("tracks", conn, if_exists="replace", index=False)

print("Fresh database created and table 'tracks' loaded.")

```

```python
create_stmt = pd.read_sql("""
    SELECT sql 
    FROM sqlite_master 
    WHERE type='table' AND name='tracks';
""", conn)

print(create_stmt.sql[0])
```

```python
pd.read_sql("SELECT COUNT(*) FROM tracks", conn)
```

```python
query_genres = """
SELECT genre, COUNT(*) AS count, AVG(popularity) AS avg_popularity
FROM tracks
GROUP BY genre
ORDER BY count DESC
LIMIT 20;
"""
pd.read_sql(query_genres, conn)
```

```python
query_years = """
SELECT year, COUNT(*) AS count, AVG(popularity) AS avg_popularity
FROM tracks
GROUP BY year
ORDER BY year;
"""
pd.read_sql(query_years, conn)
```

```python
query_summary = """
SELECT
    MIN(popularity) AS min_pop,
    MAX(popularity) AS max_pop,
    AVG(popularity) AS avg_pop
FROM tracks;
"""
pd.read_sql(query_summary, conn)
```
